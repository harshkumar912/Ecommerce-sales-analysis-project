```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load CSV with explicit column names
cols = ['OrderID', 'CustomerName', 'City', 'OrderDate', 'Category', 'Amount']
df = pd.read_csv('ecommerce_sales_data.csv', names=cols, header=None)

# Data Overview
print("Dataset Head:")
print(df.head())
print("\nDataset Info:")
print(df.info())

# 2. Visual 1: Revenue by City
plt.figure(figsize=(8, 4))
city_sales = df.groupby('City')['Amount'].sum().reset_index().sort_values(by='Amount', ascending=False)
sns.barplot(data=city_sales, x='City', y='Amount', palette='viridis')
plt.title('Total Revenue by City')
plt.xlabel('City')
plt.ylabel('Revenue (INR)')
plt.tight_layout()
plt.savefig('city_revenue_trend.png')
plt.show()

# 3. Visual 2: Category Revenue Breakdown
plt.figure(figsize=(6, 6))
cat_sales = df.groupby('Category')['Amount'].sum()
plt.pie(cat_sales, labels=cat_sales.index, autopct='%1.1f%%', startangle=140)
plt.title('Revenue Contribution by Category')
plt.tight_layout()
plt.savefig('category_distribution.png')
plt.show()
```
